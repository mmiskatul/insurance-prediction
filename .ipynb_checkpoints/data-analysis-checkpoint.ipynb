```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv('insurance.csv')
```

```python
df.head()
```

```python
df['occupation'].unique()
```

```python
df_feat =df.copy()
```

```python
df_feat['bmi']=df['weight']/(df['height']*df['height'])
```

```python
df_feat.head()
```

```python
# Feature 2: Age Group
def age_group(age):
    if age < 25:
        return "young"
    elif age < 45:
        return "adult"
    elif age < 60:
        return "middle_aged"
    return "senior"
```

```python
df_feat['age_group'] =df_feat['age'].apply(age_group)
```
